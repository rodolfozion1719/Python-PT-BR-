```python
#Normalização de dados
#Idade (0 até 100)
#Altura (1.50 até 2.00)

#formula da normalização
#(x - min) / (max - min)

dados = [10, 1, 30, 45, 50, 60, 60] #10 é o menor valor e 60 é o maior valor
min_valor = min(dados)
max_valor = max(dados)
print("Valor Mínimo: ", min_valor)
print("Valor Máximo: ", max_valor)

normalizados = []
for valor in dados:
  norm = (valor - min_valor) / (max_valor - min_valor)
  #normalizados.append(norm)#sem arredondamento
  normalizados.append(round(norm,3))#função para arredondar

print(dados)
print(normalizados)
```

Crie um programa que:

* Solicite 5 números ao usuário e armazene em um vetor
* Mostre o maior, o menor e a média

```python
numeros = []

#Entrada de dados
for i in range(5):
  num = float(input(f"Digite o {i+1}º número: "))
  numeros.append(num)

#Processamento
maior = max(numeros)
menor = min(numeros)
media = sum(numeros) / len(numeros)
ordenado = sorted(numeros)

#Saída
print("Vetor não ordenado: ", numeros)
print("Maior valor: ", maior)
print("Menor valor: ", menor)
print("Média: ", media)
print("Vetor ordenado: ", ordenado)
```

Crie um programa que:
* Gere um vetor de 8 posições (Tamanho Fixo)
* Leia 8 números inteiros e armazene no vetor
* Calcule a soma de TODOS os elementos (sem usar sum())
* Calcule a média aritmética (sem usar len())
* Mostre quantos números estão acima da média

```python
TAMANHO = 8
vetor = [0] * TAMANHO

#Preenchedo o vetor
for i in range(TAMANHO):
  vetor[i] = int(input(f"Digite o {i+1}º número: "))

#Calculando a Soma
soma = 0
for i in range(TAMANHO):
   soma += vetor[i]

#Calculando a média
media = soma / TAMANHO

#Contando QTD de números acima da média
acima = 0
for i in range(TAMANHO):
  if vetor[i] > media:
    acima = acima + 1

print(f"Vetor: {vetor}")
print(f"Soma: {soma}")
print(f"Média: {media:.2f}")
print(f"A QTD de números acima da média é: {acima}")
```

```python
TAMANHO = 15
notas = [0] * TAMANHO

#Atribui as notas ao vetor
print(f"Digite as {TAMANHO} notas dos alunos")
for i in range(TAMANHO):
  notas[i] = float(input(f"Nota do aluno {i+1}: "))

#Calcula a média
soma = 0
for i in range(TAMANHO):
   soma = soma + notas[i]
media = soma / TAMANHO

#QTD acima e abaixo da média
acima = 0
abaixo = 0
maior = notas[0]
menor = notas[0]

for i in range(TAMANHO):
   if notas[i] > media:
      acima = acima + 1
   elif notas[i] < media:
      abaixo = abaixo + 1

   if notas[i] > maior:
      maior = notas[i]
   elif notas[i] < menor:
      menor = notas[i]

print(f"Média da turma: {media:.2f}")
print(f"Alunos acima da média: {acima}")
print(f"Alunos abaixo da média: {abaixo}")
print(f"Maior nota: {maior}")
print(f"Menor nota: {menor}")
```

```python
#ler_notas.py
def ler_notas(caminho):
  notas = []
  with open(caminho, "r", encoding="utf-8") as f:
    linhas = f.readlines()
  for linha in linhas[1:]:
    campos = linha.strip.split(",")
    notas.append(float(campos[4]))
  return notas
```

```python
class Filme():
  def __init__(self, titulo, descricao, arquivo): #Primeiro método, método construtor, que a classe vai ter, com os atributos que a classe vai ter.
  #O self é para falar que é para o atributo do que estamos criando.
    self.titulo = titulo #Atributo - Características do filme
    self.descricao = descricao #Atributo
    self.arquivo = arquivo #Atributo

  def assistir_filme(self): #Segundo método
    duracao = self.calcular_duracao()
    print("Dar play no filme com o arquivo:", self.arquivo)
    print(f"O filme tem duração de {duracao} minutos")

  def mostrar_info(self): #Terceiro método e assim vai...
    print("Título:", self.titulo)
    print("Descrição:", self.descricao)
    print("Arquivo:", self.arquivo)

  def calcular_duracao(self):
    return 150

filme1 = Filme("O conde de monte cristo", "filme de ação", "conde.mp4")
filme2 = Filme("Titanic", "filme de romance", "titanic.mp4")

filme1.assistir_filme()
filme1.mostrar_info()

print("\n")

filme2.assistir_filme()
filme2.mostrar_info()

```

```python
import math

class Circulo():
  def __init__(self, raio):
    self.raio = raio

  def calcular_area(self):
    return math.pi*self.raio*self.raio

  def calcular_perimetro(self):
    return 2*math.pi*self.raio

  def mostrar(self):
    area = self.calcular_area()
    perimetro = self.calcular_perimetro()
    print(f"A área é {area} e o perímetro é {perimetro}")

c1 = Circulo()
c1.raio = 5

resultado = c1.mostrar()
print(resultado)
```

```python
class Instrumento:
    def __init__(self, nome, tipo):
        self.nome = nome
        self.tipo = tipo

    def som(self):
        return "som genérico"


class Violao(Instrumento):
    def __init__(self, tipo, estilo_corda="dedilhar cordas"):
        super().__init__(nome="Violão", tipo=tipo)
        self.estilo_corda = estilo_corda

    def som(self):
        return "som de violão acústico"


class Piano(Instrumento):
    def __init__(self, tipo, estilo_toque="clássico"):
        super().__init__(nome="Piano", tipo=tipo)
        self.estilo_toque = estilo_toque

    def som(self):
        return "som de piano de cauda"


class Bateria(Instrumento):
    def __init__(self, tipo, estilo_batida="rítmica"):
        super().__init__(nome="Bateria", tipo=tipo)
        self.estilo_batida = estilo_batida

    def som(self):
        return "som de bateria e pratos"


# Instanciação correta
orquestra = [
    Violao("Nylon"),
    Piano("Cauda"),
    Bateria("Folk")
]

# Polimorfismo em ação
for instrumento in orquestra:
    print(f"{instrumento.nome} ({instrumento.tipo}): {instrumento.som()}")
```

```python
import tkinter as tk
from tkinter import ttk, messagebox


class AppCadastro:
    def __init__(self, janela):
        self.janela = janela
        self.janela.title("Janela de cadastro")
        self.janela.geometry("350x300")

        ttk.Label(janela, text="Nome:").pack(pady=5)
        self.entry_nome = ttk.Entry(janela)
        self.entry_nome.pack()

        ttk.Label(janela, text="Idade:").pack(pady=5)
        self.entry_idade = ttk.Entry(janela)
        self.entry_idade.pack()

        ttk.Button(janela, text="Salvar", command=self.salvar_dados).pack(pady=10)

        self.text_saida = tk.Text(janela, height=6, width=40)
        self.text_saida.pack(pady=5)

    def salvar_dados(self):
        nome = self.entry_nome.get()
        idade_str = self.entry_idade.get()

        if not nome.strip():
            messagebox.showwarning("Aviso!", "O campo do nome não foi preenchido")
            return

        try:
            idade = int(idade_str)
            if idade < 0:
                raise ValueError("A idade não pode ser negativa.")
            self.text_saida.insert(tk.END, f"Cadastrado: {nome} ({idade} anos)\n")
            messagebox.showinfo("Sucesso", f"{nome} foi cadastrado com sucesso!")
            self.entry_nome.delete(0, tk.END)
            self.entry_idade.delete(0, tk.END)

        except ValueError as e:
            # int("abc") gera uma mensagem em inglês; trocamos por uma amigável
            msg = str(e) if "negativa" in str(e) else "Digite um número inteiro válido para a idade."
            messagebox.showerror("Erro", msg)


if __name__ == "__main__":
    janela = tk.Tk()
    app = AppCadastro(janela)
    janela.mainloop()
```

```python
class ContaBancaria:
    def __init__(self,titular,saldo=0.0):
        self._titular = titular
        self._saldo = saldo

    def get_titular(self):
        return self._titular

    def get_saldo(self):
        return self._saldo

    def set_titular(self,novo):
        if novo != "":
            self._titular = novo
        else:
            print("O titular não pode ser vazio")

    def set_saldo(self,novo):
        if novo != "":
            self._saldo >= 0
        else:
            print("O saldo não pode ser vazio")

    def depositar(self, valor):
        if valor > 0:
            self._saldo += valor
        else:
            print("O valor do depósito deve ser positivo.")

    def sacar(self,valor):
        if 0 < valor <= self._saldo:
            self._saldo = self._saldo - valor
        else:
            print("Saldo insuficiente ou valor inválido.")

    def __str__(self):
        return f"O titular é {self._titular} e o saldo é {self._saldo}"

class ContaPoupanca(ContaBancaria):
    def __init__(self,titular="",saldo=0.0,aniversario=""):
        super().__init__(titular,saldo)
        self._aniversario = aniversario


    def get_aniversario(self):
        return self._aniversario


    def set_aniversario(self,novo):
        if novo != "":
            self._aniversario = novo
        else:
            print("O aniversario não pode ser vazio")

    def aplicar_rendimento(self,taxa):
        self._saldo += self._saldo *(taxa/100)

cp= ContaPoupanca("Rodolfo", 1000, 22)
print("Conta criada:", cp)

cp.depositar(500.0)
print("Após depósito:", cp)

cp.aplicar_rendimento(1)
print("Após rendimento de 1%:", cp)
```

```python
from abc import ABC, abstractmethod
class Esportes(ABC):
    def __init__ (self,nome=""):
        self.nome = nome

    @abstractmethod
    def jogar():
        pass
    def __str__(self):
        return f"Estou jogando um esporte aleatório"

class Futebol(Esportes):
    def __init__(self,nome=""):
        super().__init__(nome)

    def jogar(self):
        return "Com os pés"

class Basquete(Esportes):
    def __init__(self,nome=""):
        super().__init__(nome)

    def jogar(self):
        return "Com as mãos"

class Tenis(Esportes):
    def __init__(self,nome=""):
        super().__init__(nome)

    def jogar(self):
        return "Com a raquete"

campeonato = [Futebol("Futebol"), Basquete("Basquete"), Tenis("Tenis")]
for e in campeonato:
    print(f"{e.nome} : {e.jogar()}")
```

```python
class Produto:
    def __init__(self, nome="", preco=0.0, quantidade=0.0):
        self._nome = nome
        self._preco = preco
        self._quantidade = quantidade

    # --- Property Nome (getter + setter) ---
    @property
    def nome(self):
        return self._nome

    @nome.setter
    def nome(self, novo):
        if novo != "":
            self._nome = novo
        else:
            print("Nome não pode ser vazio.")

    # --- Property Preco (getter + setter que recusa preco <= 0) ---
    @property
    def preco(self):
        return self._preco

    @preco.setter
    def preco(self, novo):
        if novo > 0:
            self._preco = novo
        else:
            print("Preço deve ser maior que zero.")

    # --- Property Quantidade (somente leitura: sem setter) ---
    @property
    def quantidade(self):
        return self._quantidade

    # --- Método vender ---
    def vender(self, qtd):
        if qtd <= self._quantidade:
            self._quantidade -= qtd
            return "Venda ok"
        else:
            return "Estoque insuficiente"

    # --- Método __str__ ---
    def __str__(self):
        return f"Produto: {self._nome}, Preço: R$ {self._preco:.2f}, Quantidade: {self._quantidade}"

    # 1. Cria o produto
p = Produto("Bola", 25.0, 10)

# 2. Imprime p.nome
print(p.nome)

# 3. Tenta atribuir valores inválidos via setter
p.nome = ""
p.preco = 0

# 4. Testa vendas (estoque suficiente vs insuficiente)
print(p.vender(3))
print(p.vender(99))

# 5. Imprime p.preco, p.quantidade e o objeto p
print(p.preco)
print(p.quantidade)
print(p)
```

```python
import tkinter as tk

class Janela(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Contador de clique")
        self.geometry("600x300")

        # Atributo da classe para guardar a contagem
        self.cliques = 0

        # 1. Criando os widgets (Composição: passam 'self' como dono)
        self.texto_contador = tk.Label(self, text="Cliques: 0", font=("Arial", 20))
        self.botao = tk.Button(self, text="Clique aqui!", font=("Arial", 14), command=self.contar)

        # 2. Posicionando na tela (empilhando um abaixo do outro)
        self.texto_contador.pack(pady=30) # pady dá um respiro/margem vertical
        self.botao.pack(pady=10)

    # 3. Método acionado toda vez que o botão for clicado
    def contar(self):
        self.cliques += 1
        # .config() atualiza o texto do Label na tela em tempo real
        self.texto_contador.config(text=f"Cliques: {self.cliques}")

janela = Janela()
janela.mainloop()
```

```python
import numpy as np
import pandas as pd

dados = {
    "time": ["Flamengo", "Palmeiras", "Atlético-MG", "Fluminense",
             "Corinthians", "Grêmio", "São Paulo", "Cruzeiro"],
    "v":    [12, 11, 10, 9, 7, 6, 5, 4],
    "e":    [4, 5, 4, 5, 6, 5, 7, 4],
    "d":    [3, 3, 5, 5, 6, 8, 7, 11],
    "gp":   [34, 30, 28, 25, 22, 20, 18, 14],
    "gc":   [15, 14, 20, 22, 23, 26, 24, 30],
}
df = pd.DataFrame(dados)
#df = pd.read_csv("brasileirao_dados.csv")
df
```

```python
df["saldo"] = df["gp"] - df["gc"]
df["saldo_pos"] = (df["saldo"] >=0).astype(int) # 1 se for >= 0, e 0 senão
print(df[["time", "saldo", "saldo_pos"]])
```

```python
#Separar em features(X) e target(y)
X = df[["v","e","d","gp","gc"]]

y = df["saldo_pos"]

print("Features shape:", X.shape)
print("Features shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

#Dividir a base em 70% e 30%
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.3, random_state = 42
)
print("Treino:", X_train.shape[0], "linhas")
print("Teste: ", X_test.shape[0], "linhas")
```

```python
from sklearn.neighbors import KNeighborsClassifier

#1. Criar o modelo (k = 3)
modelo = KNeighborsClassifier(n_neighbors = 3)

#2. Treinar no conjunto de teste
modelo.fit(X_train, y_train)

print("Modelo Treinado")
```

```python
y_pred = modelo.predict(X_test)

print("Previsões do modelo: ", y_pred)
print("Valores Reais: ", y_test.values)
```

```python
from sklearn.metrics import accuracy_score
#Acurácia no teste
acuracia = modelo.score(X_test,y_test)
print(f"Acurácia no teste: {acuracia:.2%}")

#Acurácia na previsão
acuracia = accuracy_score(y_test, y_pred)
print("Acurácia:", acuracia)
print("Ou: ", f"{acuracia * 100:.1f}%")
```
